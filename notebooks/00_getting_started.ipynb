# Phi-nance in JupyterLab

Run the next cell **once per session**. It locates the repo, adds it to `sys.path`, sets `IS_BACKTESTING=True` (required before importing lumibot), and loads `.env`.

Use kernel **Python (Phi-nance)** if you registered it (`pip install -r requirements-jupyter.txt` then `python -m ipykernel install --user --name phinance --display-name "Python (Phi-nance)"`).

```python
import runpy
from pathlib import Path


def _find_notebook_setup() -> Path:
    for base in [Path.cwd(), *Path.cwd().parents]:
        p = base / "notebook_setup.py"
        if p.is_file():
            return p
    raise FileNotFoundError(
        "notebook_setup.py not found. Open JupyterLab with the Phi-nance folder as the workspace."
    )


_ns = runpy.run_path(str(_find_notebook_setup()), run_name="__main__")
globals().update({k: _ns[k] for k in ("ROOT", "find_project_root", "setup") if k in _ns})
print("ROOT =", ROOT)
```
